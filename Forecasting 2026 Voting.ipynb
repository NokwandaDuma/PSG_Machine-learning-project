```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
# ==========================================
# LOAD THE THREE ELECTION DATASETS
# ==========================================

df_2024 = pd.read_excel("2024 National & Provincial Elections.xls", sheet_name="Results Report - Municipal", header=None)
df_2016 = pd.read_excel("local goverment elections 2016.xls", sheet_name="Results Summary", header=None)
df_2021 = pd.read_excel("local goverment elections 2021.xls", sheet_name="Results Summary", header=None)

print("2024 shape:", df_2024.shape)
print("2016 shape:", df_2016.shape)
print("2021 shape:", df_2021.shape)
```

```python
# ==========================================
# 2024 NATIONAL ELECTION - REGIONAL BALLOT (eThekwini)
# Party rows start after the header row containing "Party Name"
# ==========================================

# Find the header row for 2024
header_2024 = None
for i in range(len(df_2024)):
    if "Party Name" in str(df_2024.iloc[i, 1]):
        header_2024 = i
        break

print("2024 header row index:", header_2024)

# Extract party rows (until "Total Valid Votes")
parties_2024 = []
for i in range(header_2024 + 1, len(df_2024)):
    party_name = df_2024.iloc[i, 1]
    abbr = df_2024.iloc[i, 14]
    votes = df_2024.iloc[i, 16]
    pct = df_2024.iloc[i, 17]
    if pd.isna(party_name) or "Total" in str(party_name):
        break
    parties_2024.append({
        "Party": party_name,
        "Abbr": abbr,
        "Votes_2024": votes,
        "Pct_2024": pct
    })

df_parties_2024 = pd.DataFrame(parties_2024)
df_parties_2024 = df_parties_2024.dropna(subset=["Votes_2024"])
df_parties_2024["Votes_2024"] = pd.to_numeric(df_parties_2024["Votes_2024"], errors="coerce")
df_parties_2024["Pct_2024"] = pd.to_numeric(df_parties_2024["Pct_2024"], errors="coerce")

print(df_parties_2024.head(10))
print("Total parties 2024:", len(df_parties_2024))
print("Total valid votes 2024:", df_parties_2024["Votes_2024"].sum())
```

```python
# ==========================================
# 2016 LOCAL GOVERNMENT ELECTION (eThekwini)
# ==========================================

header_2016 = None
for i in range(len(df_2016)):
    if "Party Name" in str(df_2016.iloc[i, 1]):
        header_2016 = i
        break

print("2016 header row index:", header_2016)

parties_2016 = []
for i in range(header_2016 + 2, len(df_2016)):
    party_name = df_2016.iloc[i, 1]
    ward_votes = df_2016.iloc[i, 4]
    pr_votes = df_2016.iloc[i, 8]
    total_votes = df_2016.iloc[i, 11]
    total_pct = df_2016.iloc[i, 12]
    if pd.isna(party_name) or "Total" in str(party_name):
        break
    parties_2016.append({
        "Party": party_name,
        "Ward_Votes_2016": ward_votes,
        "PR_Votes_2016": pr_votes,
        "Total_Votes_2016": total_votes,
        "Pct_2016": total_pct
    })

df_parties_2016 = pd.DataFrame(parties_2016)
df_parties_2016 = df_parties_2016.dropna(subset=["Total_Votes_2016"])
df_parties_2016["Total_Votes_2016"] = pd.to_numeric(df_parties_2016["Total_Votes_2016"], errors="coerce")
df_parties_2016["Pct_2016"] = pd.to_numeric(df_parties_2016["Pct_2016"], errors="coerce")

print(df_parties_2016.head(10))
print("Total parties 2016:", len(df_parties_2016))
print("Total valid votes 2016:", df_parties_2016["Total_Votes_2016"].sum())
```

```python
# ==========================================
# 2021 LOCAL GOVERNMENT ELECTION (eThekwini)
# ==========================================

header_2021 = None
for i in range(len(df_2021)):
    if "Party Name" in str(df_2021.iloc[i, 1]):
        header_2021 = i
        break

print("2021 header row index:", header_2021)

parties_2021 = []
for i in range(header_2021 + 2, len(df_2021)):
    party_name = df_2021.iloc[i, 1]
    ward_votes = df_2021.iloc[i, 4]
    pr_votes = df_2021.iloc[i, 8]
    total_votes = df_2021.iloc[i, 11]
    total_pct = df_2021.iloc[i, 12]
    if pd.isna(party_name) or "Total" in str(party_name):
        break
    parties_2021.append({
        "Party": party_name,
        "Ward_Votes_2021": ward_votes,
        "PR_Votes_2021": pr_votes,
        "Total_Votes_2021": total_votes,
        "Pct_2021": total_pct
    })

df_parties_2021 = pd.DataFrame(parties_2021)
df_parties_2021 = df_parties_2021.dropna(subset=["Total_Votes_2021"])
df_parties_2021["Total_Votes_2021"] = pd.to_numeric(df_parties_2021["Total_Votes_2021"], errors="coerce")
df_parties_2021["Pct_2021"] = pd.to_numeric(df_parties_2021["Pct_2021"], errors="coerce")

print(df_parties_2021.head(10))
print("Total parties 2021:", len(df_parties_2021))
print("Total valid votes 2021:", df_parties_2021["Total_Votes_2021"].sum())
```

```python
# ==========================================
# STANDARDISE PARTY NAMES FOR CROSS-YEAR MERGING
# ==========================================

def standardise_party(name):
    if pd.isna(name):
        return name
    name = str(name).upper().strip()
    name = name.replace("  ", " ")
    # Common aliases
    aliases = {
        "AFRICAN NATIONAL CONGRESS": "ANC",
        "DEMOCRATIC ALLIANCE": "DA",
        "INKATHA FREEDOM PARTY": "IFP",
        "ECONOMIC FREEDOM FIGHTERS": "EFF",
        "UNITED DEMOCRATIC MOVEMENT": "UDM",
        "PAN AFRICANIST CONGRESS OF AZANIA": "PAC",
        "AFRICAN CHRISTIAN DEMOCRATIC PARTY": "ACDP",
        "VRYHEIDSFRONT PLUS": "VF PLUS",
        "CONGRESS OF THE PEOPLE": "COPE",
        "AFRICAN INDEPENDENT CONGRESS": "AIC",
        "AL JAMA-AH": "AL JAMA-AH",
        "MINORITY FRONT": "MF",
        "AFRICAN TRANSFORMATION MOVEMENT": "ATM",
        "ACTIONSA": "ACTIONSA",
        "GOOD": "GOOD",
        "PATRIOTIC ALLIANCE": "PA",
        "NATIONAL FREEDOM PARTY": "NFP",
        "AFRICAN PEOPLE'S CONVENTION": "APC",
        "AZANIAN PEOPLE'S ORGANISATION": "AZAPO",
        "ALLIED MOVEMENT FOR CHANGE": "AM4C",
        "DEMOCRATIC LIBERAL CONGRESS": "DLC",
        "UNITED INDEPENDENT MOVEMENT": "UIM",
        "AFRICAN MANTUNGWA COMMUNITY": "AMC",
        "MINORITIES OF SOUTH AFRICA": "MOSA",
        "TRULY ALLIANCE": "TA",
        "PEOPLE'S REVOLUTIONARY MOVEMENT": "PRM",
        "INDEPENDENT": "INDEPENDENT",
    }
    return aliases.get(name, name)

df_parties_2024["Party_Std"] = df_parties_2024["Party"].apply(standardise_party)
df_parties_2016["Party_Std"] = df_parties_2016["Party"].apply(standardise_party)
df_parties_2021["Party_Std"] = df_parties_2021["Party"].apply(standardise_party)
```

```python
print("2024 duplicates:", df_parties_2024.duplicated(subset=["Party_Std"]).sum())
print("2016 duplicates:", df_parties_2016.duplicated(subset=["Party_Std"]).sum())
print("2021 duplicates:", df_parties_2021.duplicated(subset=["Party_Std"]).sum())
```

```python
df_2016_before = df_parties_2016.copy()

print("Before removing duplicates:", df_parties_2016.shape)
print("Rows involved in duplicates:", df_parties_2016.duplicated(subset=["Party_Std"], keep=False).sum())
print("Rows that will be removed:", df_parties_2016.duplicated(subset=["Party_Std"]).sum())
```

```python
# Aggregate duplicates (sum votes) rather than dropping
df_parties_2016 = df_parties_2016.groupby("Party_Std", as_index=False).agg({
    "Party": "first",
    "Total_Votes_2016": "sum",
    "Ward_Votes_2016": "sum",
    "PR_Votes_2016": "sum"
})

df_parties_2021 = df_parties_2021.groupby("Party_Std", as_index=False).agg({
    "Party": "first",
    "Total_Votes_2021": "sum",
    "Ward_Votes_2021": "sum",
    "PR_Votes_2021": "sum"
})

df_parties_2024 = df_parties_2024.groupby("Party_Std", as_index=False).agg({
    "Party": "first",
    "Votes_2024": "sum"
})

# Recompute percentages
df_parties_2016["Pct_2016"] = df_parties_2016["Total_Votes_2016"] / df_parties_2016["Total_Votes_2016"].sum()
df_parties_2021["Pct_2021"] = df_parties_2021["Total_Votes_2021"] / df_parties_2021["Total_Votes_2021"].sum()
df_parties_2024["Pct_2024"] = df_parties_2024["Votes_2024"] / df_parties_2024["Votes_2024"].sum()

print("2016 shape after aggregation:", df_parties_2016.shape)
print("2021 shape after aggregation:", df_parties_2021.shape)
print("2024 shape after aggregation:", df_parties_2024.shape)
```

```python
# ==========================================
# MERGE ACROSS YEARS
# ==========================================

df_master = df_parties_2016[["Party_Std", "Party", "Total_Votes_2016", "Pct_2016"]].merge(
    df_parties_2021[["Party_Std", "Total_Votes_2021", "Pct_2021"]],
    on="Party_Std",
    how="outer"
).merge(
    df_parties_2024[["Party_Std", "Votes_2024", "Pct_2024"]],
    on="Party_Std",
    how="outer"
)

# Fill missing votes with 0 (party did not contest that year)
df_master["Total_Votes_2016"] = df_master["Total_Votes_2016"].fillna(0)
df_master["Total_Votes_2021"] = df_master["Total_Votes_2021"].fillna(0)
df_master["Votes_2024"] = df_master["Votes_2024"].fillna(0)

df_master["Pct_2016"] = df_master["Pct_2016"].fillna(0)
df_master["Pct_2021"] = df_master["Pct_2021"].fillna(0)
df_master["Pct_2024"] = df_master["Pct_2024"].fillna(0)

# Clean up party names
df_master["Party"] = df_master["Party"].fillna(df_master["Party_Std"])

print(df_master.shape)
print(df_master.head(20))
```

```python
# ==========================================
# FEATURES AND TARGET
# ==========================================

# Filter to parties with meaningful presence (>= 0.1% in at least one year)
df_model = df_master[
    (df_master["Pct_2016"] >= 0.001) |
    (df_master["Pct_2021"] >= 0.001) |
    (df_master["Pct_2024"] >= 0.001)
].copy()

print("Parties included for modelling:", len(df_model))

# Features: historical vote shares
X = df_model[["Pct_2016", "Pct_2021"]]

# Target: 2024 vote share (as proxy for 2026 projection)
y = df_model["Pct_2024"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0),
    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        max_depth=5,
        min_samples_split=3,
        random_state=42,
        n_jobs=-1
    ),
    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=200,
        max_depth=3,
        learning_rate=0.05,
        random_state=42
    )
}
```

```python
results = []

for name, model in models.items():
    print(f"\nTraining {name}...")
    model.fit(X_train, y_train)

    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    train_r2 = r2_score(y_train, train_pred)
    test_r2 = r2_score(y_test, test_pred)
    test_mae = mean_absolute_error(y_test, test_pred)

    results.append({
        "Model": name,
        "Train R2": train_r2,
        "Test R2": test_r2,
        "Test MAE": test_mae
    })

    print(f"  Train R2: {train_r2:.4f}")
    print(f"  Test R2:  {test_r2:.4f}")
    print(f"  Test MAE: {test_mae:.4f}")
```

```python
results_df = pd.DataFrame(results)
print("\n===== MODEL RESULTS =====")
print(results_df.round(4))
```

```python
# ==========================================
# PROJECT 2026 VOTE SHARES
# ==========================================

# Use the best model (Random Forest by default)
best_model = models["Random Forest"]

# Use 2021 and 2024 as the most recent historical features
X_2026 = df_model[["Pct_2021", "Pct_2024"]].copy()

# Rename columns to match training features
X_2026.columns = ["Pct_2016", "Pct_2021"]

# Predict
df_model["Pct_2026_Predicted"] = best_model.predict(X_2026)

# Normalise to sum to 1
df_model["Pct_2026_Predicted"] = (
    df_model["Pct_2026_Predicted"] /
    df_model["Pct_2026_Predicted"].sum()
)

# Show top projected parties
projected_top = df_model.sort_values("Pct_2026_Predicted", ascending=False).head(10)
print(projected_top[["Party_Std", "Pct_2021", "Pct_2024", "Pct_2026_Predicted"]])
```

```python
# ==========================================
# PROJECTED VOTER TURNOUT
# ==========================================

# Historical turnout (approximate)
turnout_2016 = total_valid_2016 / registered_2024  # using 2024 registration as baseline
turnout_2021 = total_valid_2021 / registered_2024
turnout_2024 = total_valid_2024 / registered_2024

print(f"2016 Turnout (approx): {turnout_2016:.2%}")
print(f"2021 Turnout (approx): {turnout_2021:.2%}")
print(f"2024 Turnout (approx): {turnout_2024:.2%}")

# Project 2026 turnout using linear trend
years = np.array([2016, 2021, 2024]).reshape(-1, 1)
turnouts = np.array([turnout_2016, turnout_2021, turnout_2024])

turnout_model = LinearRegression()
turnout_model.fit(years, turnouts)

turnout_2026 = turnout_model.predict([[2026]])[0]

print(f"\nProjected 2026 Turnout Rate: {turnout_2026:.2%}")
print(f"Projected 2026 Votes Cast: {int(turnout_2026 * registered_2024):,}")
```

```python
# ==========================================
# PROJECTED 2026 VOTE TOTALS
# ==========================================

projected_votes_2026 = int(turnout_2026 * registered_2024)

df_model["Projected_Votes_2026"] = (
    df_model["Pct_2026_Predicted"] * projected_votes_2026
).round(0).astype(int)

projected_votes_top = df_model.sort_values("Pct_2026_Predicted", ascending=False).head(10)

print("Projected 2026 Vote Totals (Top 10):")
print(projected_votes_top[["Party_Std", "Pct_2026_Predicted", "Projected_Votes_2026"]])
```
